```python
import matplotlib.pyplot as plt  
x = ["python","c","c++","java"]
y = [85,65,72,90]

plt.xlabel("language",fontsize = 15)
plt.ylabel("No.")
plt.title("wscube")
c = ["y","b","g","r"]
plt.bar(x,y,width = 0.4, color = c, edgecolor = "r", linewidth = 5)
plt.show()
```

```python
import matplotlib.pyplot as plt  
import numpy as np
x = ["python","c","c++","java"]
y = [85,65,72,90]
z = [20,30,40,50]

width = 0.2
p = np.arange(len(x))
p1 = [j+width for j in p]
plt.xlabel("language",fontsize = 15)
plt.ylabel("No.")
plt.title("wscube")

plt.bar(p,y,width , color = "b", label = "popularity")
plt.bar(p1,z,width, color = "y",label = "popularity1")

plt.xticks(p+width/2,x)
plt.legend()
plt.show()
```
