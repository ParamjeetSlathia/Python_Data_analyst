```python
import matplotlib.pyplot as plt

x = [25,35,45,75]
y = [10,20,30,40]

plt.plot(x,y)
plt.show()
```

```python
# customize chart

plt.figure(figsize = (4,3))

plt.plot(x,y, color = 'g', marker = 'o', linestyle = 'dashed', linewidth = 2, markersize = 12)

plt.title("kuch bii title do")
plt.xlabel("x axis label ha")
plt.ylabel("y axis label ha")

plt.show()
```

```python
# advance line chart

x = [25,35,45,75]
y1 = [10,20,30,40]
y2 = [20,30,40,50]

plt.plot(x,y1, label = 'sales 2024')
plt.plot(x,y2, label = 'sales 2025')

plt.title('sales data')
plt.xlabel('month')
plt.ylabel('sales')

plt.legend()
plt.show()
```

```python
# bar chart

y = [25,35,45,75]
x = [10,20,30,40]

plt.bar(x,y)
plt.title("bar chart")
plt.show()
```

```python
# histogram -  used for distribution analysis

import random

data = [random.randint(1,20) for i in range(500)]

plt.hist(data, bins = 20)
plt.title("histogram example")
plt.show()
```

```python
# pie chart -  used to show part -to- whole relationships/data

categories = ['A','B','C','D','E']
sales = [10,20,55,35,45]

plt.pie(sales, labels = categories, autopct='%1.1f%%')
plt.title("Pie chart example")
plt.show()
```

```python
# scatter plot - used to find relationship between variables

y1 = [10,20,30,40]
y2 = [20,30,40,50]

plt.scatter(y1,y2)
plt.title("scatter")
plt.show()
```

```python
# sub plot - used to show multiple chart in one figure

#data 1
categories = ['Mon','tue','web','thu','fri']
sales = [10,20,55,35,45]

# data 2
y1 = [10,20,30,40]
y2 = [20,30,40,50]

plt.figure(figsize = (10,4))

# first subplot - bar chart

plt.subplot(1,2,1) # row,column,position
plt.bar(categories, sales)
plt.title("daily sales")
plt.xlabel("week days")
plt.ylabel("sales")


# second subplot - scatter chart

plt.subplot(1,2,2)
plt.scatter(y1,y2)
plt.title("user example")
plt.xlabel("user 1")
plt.ylabel("user 2")

plt.show()
```

```python
# matplot with pandas with real data

import pandas as pd   

data = {
    'Month' : ['Jan','Feb','Mar','Apr','May'],
    'sales' : [2000,3000,4000,5000,6000]
}

df = pd.DataFrame(data)
df
    
```

```python
plt.bar(df['Month'],df['sales'])
plt.title("Matplotlib with pandas")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```
